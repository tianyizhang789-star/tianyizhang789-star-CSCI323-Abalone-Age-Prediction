```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
X_train = pd.read_csv('X_train_scaled.csv')
X_test = pd.read_csv('X_test_scaled.csv')
y_train = pd.read_csv('y_train.csv').values.flatten() + 1.5
y_test = pd.read_csv('y_test.csv').values.flatten() + 1.5
```

Load the standardized feature data that has been processed

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
```

Assuming there is a linear relationship between the data

```python
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")
```

```python
residuals = y_test - y_pred_lr
```

Residual = Actual - Predicted

```python
sns.regplot(x=y_test, y=residuals, lowess=True, 
            scatter_kws={'alpha':0.4, 'color':'steelblue', 's':30}, 
            line_kws={'color':'red', 'lw':3, 'label': 'Non-linear Trend'})

plt.axhline(0, color='black', linestyle='--', lw=1.5)
plt.title('Residual Plot of Linear Regression', fontsize=14)
plt.xlabel('Actual Age', fontsize=12)
plt.ylabel('Residuals (Actual - Predicted)', fontsize=12)
plt.legend()

plt.tight_layout()
plt.show()
```
